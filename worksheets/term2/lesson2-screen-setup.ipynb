# Lesson 2: Setting Up the Screen

Before you can draw anything interesting, you need to understand the canvas you are drawing on.
This notebook covers how to configure the turtle screen and how to place the turtle exactly where you want it.

## Step 1: Creating the Screen

The `turtle.Screen()` object controls the drawing window. Always create it before your turtle, then call `screen.setup(width, height)` to set the window size in pixels.

- **Width** is how wide the window is (left to right)
- **Height** is how tall the window is (top to bottom)

````python
import turtle

screen = turtle.Screen()
screen.setup(width=600, height=600)

t = turtle.Turtle()
t.speed(3)
````

## Step 2: The Coordinate System

The turtle screen is a coordinate grid. **(0, 0) is always at the center** of the window.

- `x` increases to the **right**, decreases to the **left**
- `y` increases **upward**, decreases downward

```
         y = +300
              |
x = -300 ----+---- x = +300
              |
         y = -300
```

On a 600 × 600 screen the edges are at **±300** in both directions.

| Corner | x | y |
|---|---|---|
| Top-left | -290 | 280 |
| Top-right | 290 | 280 |
| Bottom-right | 290 | -280 |
| Bottom-left | -290 | -280 |

Run the cell below — it draws and labels each corner so you can see the grid in action.

````python
t.clear()
t.penup()
screen.colormode(255)

t.goto(-290, 280)
t.dot(8)
t.write("top-left (-290, 280)", font=("Arial", 9, "normal"))

t.goto(290, 280)
t.dot(8)
t.write("top-right (290, 280)", font=("Arial", 9, "normal"))

t.goto(290, -280)
t.dot(8)
t.write("bottom-right (290, -280)", font=("Arial", 9, "normal"))

t.goto(-290, -280)
t.dot(8)
t.write("bottom-left (-290, -280)", font=("Arial", 9, "normal"))

t.goto(0, 0)
t.dot(8)
t.write(" center (0, 0)", font=("Arial", 9, "normal"))
````

## Step 3: The goto() Command

`t.goto(x, y)` moves the turtle directly to any coordinate on the screen.

| Command | What it does |
|---|---|
| `t.goto(0, 0)` | Move to the center |
| `t.goto(100, 0)` | Move 100 pixels to the right of center |
| `t.goto(0, -150)` | Move 150 pixels below center |
| `t.goto(-100, 100)` | Move to the upper-left area |

> **Important:** If the pen is down, `goto()` will draw a line as the turtle travels. Use `penup()` before moving and `pendown()` when you are ready to draw again.

````python
t.clear()

# Move WITHOUT drawing
t.penup()
t.goto(-100, -100)   # jump to bottom-left area
t.pendown()

# Now draw from here
t.goto(100, -100)    # draw a line to the right
t.goto(100,  100)    # draw a line up
t.goto(-100,  100)   # draw a line to the left
t.goto(-100, -100)   # draw a line back to start
````

Notice that `goto()` can draw lines directly — you don't always need `forward()` and `right()`. When you know the exact coordinates you want, `goto()` is often simpler.

## Step 4: setheading()

`t.setheading(angle)` points the turtle in a specific direction before you draw. Think of it like a compass — 0° is east and the angle increases counter-clockwise.

| Command | Effect |
|---|---|
| `t.setheading(0)` | Point right (east) |
| `t.setheading(90)` | Point up (north) |
| `t.setheading(180)` | Point left (west) |
| `t.setheading(270)` | Point down (south) |

````python
t.clear()
t.penup()
t.goto(0, 0)
t.pendown()

t.setheading(90)    # face up
t.forward(100)

t.setheading(0)     # face right
t.forward(100)

t.setheading(270)   # face down
t.forward(100)

t.setheading(180)   # face left
t.forward(100)
````
