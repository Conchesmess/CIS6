## Step 1: Set Up Your Turtle

Before you can draw anything, you need to import the turtle module and create a turtle to work with.

- `turtle.Screen()` creates the drawing window
- `turtle.Turtle()` creates your turtle (the pen)
- `speed(3)` sets how fast the turtle moves (1 = slow, 10 = fast, 0 = instant)

```python
import turtle

screen = turtle.Screen()
screen.title("My Turtle Drawing")
screen.bgcolor("white")

t = turtle.Turtle()
t.speed(3)
```

## Step 2: Basic Movement

The turtle starts in the center of the screen, facing right. The four commands you need are:

| Command | What it does |
|---|---|
| `t.forward(100)` | Move forward 100 pixels |
| `t.backward(100)` | Move backward 100 pixels |
| `t.right(90)` | Turn right 90 degrees |
| `t.left(90)` | Turn left 90 degrees |

Run the cell below and watch what happens.

```python
t.clear()          # clear any previous drawing
t.home()           # return turtle to center, facing right

t.forward(100)     # draw a line 100 pixels long
```

## Step 3: Drawing One Side at a Time

A square has **4 sides** and **4 corners**. Each corner is a 90-degree right turn.

Let's draw the square one side at a time so you can see exactly what's happening.

```python
t.clear()
t.home()

# Side 1
t.forward(100)
t.right(90)

# Side 2
t.forward(100)
t.right(90)

# Side 3
t.forward(100)
t.right(90)

# Side 4
t.forward(100)
t.right(90)
```

## Step 4: Setting the Pen Color

By default the turtle draws in black. You can change the outline color with `t.pencolor()`.

You can use:
- A color name: `"red"`, `"blue"`, `"green"`
- An RGB value: `t.pencolor(255, 0, 0)` for red

To use RGB values you first need to tell the screen to accept 0–255 ranges.

```python
screen.colormode(255)   # allow RGB values from 0 to 255

t.clear()
t.home()

t.pencolor(255, 0, 0)   # red outline
t.pensize(3)            # make the line a little thicker

for i in range(4):
    t.forward(100)
    t.right(90)
```

## Step 5: Filling the Square with Color

To fill a shape, you need to:
1. Set the fill color with `t.fillcolor()`
2. Call `t.begin_fill()` **before** you start drawing
3. Draw your shape
4. Call `t.end_fill()` **after** the shape is closed

The turtle will automatically fill in the shape when you call `end_fill()`.

```python
t.clear()
t.home()

t.pencolor(255, 0, 0)     # red outline
t.fillcolor(255, 255, 0)  # yellow fill
t.pensize(3)

t.begin_fill()            # start filling

for i in range(4):
    t.forward(100)
    t.right(90)

t.end_fill()              # finish filling
```
