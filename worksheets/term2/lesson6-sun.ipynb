# Lesson 6: Draw a Sun

This lesson has three parts we build up one at a time:
1. A large yellow dot positioned so it extends off the top of the screen
2. Sun rays radiating outward using `setheading` and `forward`/`backward`
3. A yellow-to-red gradient made by stacking dots of decreasing size

**Run the Setup cell below first**, then you can run any other cell as many times as you like.

```python
import turtle

def fresh_turtle():
    turtle.TurtleScreen._RUNNING = True
    screen = turtle.Screen()
    screen.setup(600, 500)
    screen.clear()
    screen.colormode(255)
    screen.tracer(0)
    t = turtle.Turtle()
    t.speed(0)
    t.hideturtle()
    screen.getcanvas().winfo_toplevel().lift()
    return screen, t
```

## Step 1: The `dot()` command

`t.dot(size, color)` draws a filled circle centered at the turtle's current position.
- `size` is the diameter in pixels
- `color` can be a color name string or an RGB tuple (when `colormode(255)` is on)
- The turtle does **not** move — it stays at the same spot after drawing the dot

Run this to see a few dots of different sizes:

```python
screen, t = fresh_turtle()

t.dot(200, (255, 255, 0))   # large yellow dot at center

t.penup()
t.goto(150, 0)
t.dot(80, (255, 165, 0))    # smaller orange dot to the right

screen.update()
```

## Step 2: Position the sun off the top of the screen

The default turtle screen is 600 pixels wide and 500 pixels tall, so y runs from **-250 to 250**.

If we place the sun center at `(0, 250)` — the very top edge — and use a dot with diameter 350, the radius is 175. That means:
- The **bottom** of the dot is at y = 250 − 175 = **75** (visible)
- The **top** extends to y = 250 + 175 = **425** (off screen)

This makes it look like a sun rising above the horizon.

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250

t.penup()
t.goto(SUN_X, SUN_Y)
t.dot(350, (255, 255, 0))

screen.update()screen.getcanvas().winfo_toplevel().lift()
```

## Step 3: Drawing one ray

A ray is just a line going outward from the sun center and coming back. The turtle:
1. Moves to the sun center (pen up)
2. Points in a direction with `setheading()`
3. Draws forward, then backward to return to center

Run this to see a single ray pointing to the right:

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250

# Draw the sun dot
t.penup()
t.goto(SUN_X, SUN_Y)
t.dot(350, (255, 255, 0))

# Draw one ray pointing right
t.penup()
t.goto(SUN_X, SUN_Y)
t.setheading(0)
t.pendown()
t.forward(200)
t.backward(200)
t.penup()

screen.update()
```

## Step 4: Rays in a loop

To make rays all the way around we use a loop that increases the heading by a fixed amount each time. If we go from 0 to 360 in steps of 20 we get 18 rays.

Notice we **don't** need to go back to the sun center each time — `backward()` already returns us there.

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250

# Sun dot
t.penup()
t.goto(SUN_X, SUN_Y)
t.dot(350, (255, 255, 0))

# Rays
t.penup()
t.goto(SUN_X, SUN_Y)
t.pendown()
for angle in range(0, 360, 20):
    t.setheading(angle)
    t.forward(200)
    t.backward(200)

screen.update()
```

## Step 5: Extra credit — ray colors

We can change the pen color each ray using `t.pencolor()`. Here we cycle through a list of warm colors:

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250
ray_colors = [(255, 255, 0), (255, 165, 0), (255, 0, 0), (255, 165, 0)]

t.penup()
t.goto(SUN_X, SUN_Y)
t.dot(350, (255, 255, 0))

t.penup()
t.goto(SUN_X, SUN_Y)
t.pendown()
for i, angle in enumerate(range(0, 360, 20)):
    t.pencolor(ray_colors[i % len(ray_colors)])
    t.setheading(angle)
    t.forward(200)
    t.backward(200)

screen.update()
```

## Step 6: Gradient — stacking dots from yellow to red

Just like the gradient square in Lesson 4, we use `r`, `g`, `b` variables. This time:
- We start with a large yellow dot: `r=255, g=255, b=0`
- Each loop we make the dot **smaller** and subtract 1 from `g`
- After 255 loops the last (smallest) dot is red: `r=255, g=0, b=0`

The result is a sun that glows yellow at the outside and burns red at the center.

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250

r = 255
g = 255
b = 0
dot_size = 350

t.penup()
t.goto(SUN_X, SUN_Y)

for i in range(255):
    t.dot(dot_size, (r, g, b))
    dot_size -= 1
    g -= 1

screen.update()
```

## Step 7: Complete sun — rays + gradient

Now we combine everything. The order matters:
1. Draw the rays **first** (behind the sun)
2. Draw the gradient dots **on top** (covering the center of the rays)

This way the rays look like they're shooting out from behind the sun.

```python
screen, t = fresh_turtle()

SUN_X = 0
SUN_Y = 250

# --- Rays (drawn first, behind the sun) ---
t.penup()
t.goto(SUN_X, SUN_Y)
t.pencolor(255, 200, 0)
t.pendown()
for angle in range(0, 360, 20):
    t.setheading(angle)
    t.forward(200)
    t.backward(200)
t.penup()

# --- Gradient sun (drawn on top of rays) ---
r = 255
g = 255
b = 0
dot_size = 350

t.goto(SUN_X, SUN_Y)
for i in range(255):
    t.dot(dot_size, (r, g, b))
    dot_size -= 1
    g -= 1

screen.update()
```

## Challenges

1. **Move the sun:** Change `SUN_X` to place the sun to the left or right instead of center.
2. **Ray thickness:** Add `t.pensize(3)` before the ray loop to make the rays thicker.
3. **More or fewer rays:** Change the step in `range(0, 360, 20)` — a smaller step gives more rays, a larger step gives fewer.
4. **Different gradient:** Instead of yellow → red, try white → yellow by starting at `r=255, g=255, b=255` and subtracting from `b` each step.
