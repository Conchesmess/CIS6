# Lesson 4: Gradients

In this lesson we build on the solid square from Lesson 3 and make the color gradually change from **yellow** to **red** across the square.

**Run the Setup cell below first**, then you can run any other cell as many times as you like.

````python
import turtle

def fresh_turtle():
    turtle.TurtleScreen._RUNNING = True
    screen = turtle.Screen()
    screen.clear()
    screen.colormode(255)
    screen.tracer(0)
    t = turtle.Turtle()
    t.speed(0)
    t.hideturtle()
    return screen, t
````

## Step 1: Review — the Lesson 3 solid square

In Lesson 3 we drew a 255×255 square by looping 255 times, drawing a vertical line, and then stepping one pixel to the right. Run this to make sure it still works before we modify it.

````python
screen, t = fresh_turtle()

for i in range(255):
    t.penup()
    t.goto(i, 0)
    t.setheading(90)   # face upward
    t.pendown()
    t.forward(255)

screen.update()
````

## Step 2: Separate RGB variables

Instead of writing a color directly (like `'red'` or `'#ff0000'`), we will store the **red**, **green**, and **blue** channels in three separate **variables**.

**Assigning a variable** means giving a name to a value so you can use and change it later:

```python
r = 255   # r now holds the value 255
g = 255
b = 0
```

The key reason to use variables here — instead of just typing the numbers — is that the loop can **change** their values as it runs. `g` starts at 255 and we subtract 1 each line, so by the last line it has counted all the way down to 0. If we had typed the number directly we couldn't do that.

Our starting color is **yellow**, which in RGB is:
- Red = 255 (full)
- Green = 255 (full)
- Blue = 0 (none)

Our ending color is **red** (after 255 lines), which is:
- Red = 255
- Green = **0**  ← we subtract 1 per line
- Blue = 0

Run the cell below to see the starting and ending colors printed out.

````python
r = 255
g = 255
b = 0

print(f"Starting color  → r={r}, g={g}, b={b}  (yellow)")

# Simulate what happens after 255 lines
for i in range(255):
    g -= 1

print(f"Ending color    → r={r}, g={g}, b={b}  (red)")
````

## Step 3: Tell turtle to use 0–255 color values

By default, turtle uses colors as names (`'red'`) or 0–1 floats. We need to call `screen.colormode(255)` so we can pass whole numbers like `255` and `128`. The `fresh_turtle()` setup cell already handles this, so the cell below is just a quick test.

````python
screen, t = fresh_turtle()

# Draw a single line in yellow to test colormode
t.pencolor(255, 255, 0)
t.forward(100)

screen.update()
````

## Step 4: Set the pen color using r, g, b inside the loop

Now we combine the loop from Lesson 3 with the RGB variables. Before drawing each line we call `t.pencolor(r, g, b)`, then after drawing we subtract 1 from `g`.

Notice where each new line fits inside the loop:

````python
screen, t = fresh_turtle()

r = 255
g = 255
b = 0

for i in range(255):
    t.pencolor(r, g, b)   # set color BEFORE drawing

    t.penup()
    t.goto(i, 0)
    t.setheading(90)
    t.pendown()
    t.forward(255)

    g -= 1                # subtract 1 from green AFTER drawing

screen.update()
````

## Step 5: Why does the order matter?

What if we subtracted from `g` *before* drawing instead of after? The gradient would shift by one step — the first line would be slightly less yellow and the last line would be pure red minus one step.

Try swapping the `g -= 1` line to the top of the loop (before `pencolor`) and see how it changes the output.

## Step 6: Track the color as it changes

Run this cell to print the color for every 50th line so you can see the gradient in numbers:

````python
r = 255
g = 255
b = 0

for i in range(255):
    if i % 50 == 0:
        print(f"Line {i:3d}  →  rgb({r}, {g:3d}, {b})")
    g -= 1
````

## Step 7: Complete gradient — all together

Here is the final, clean version of the Lesson 4 program:

````python
screen, t = fresh_turtle()

r = 255
g = 255
b = 0

for i in range(255):
    t.pencolor(r, g, b)
    t.penup()
    t.goto(i, 0)
    t.setheading(90)
    t.pendown()
    t.forward(255)
    g -= 1

screen.update()
````
