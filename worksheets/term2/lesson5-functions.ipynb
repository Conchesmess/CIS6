# Lesson 5: Functions

In this lesson we write two functions:
- `bird(radius)` — draws one bird shape from two arcs
- `birds(num_birds)` — loops and places `bird` at random locations across the screen

**Run the Setup cell below first**, then you can run any other cell as many times as you like.

````python
import turtle

def fresh_turtle():
    turtle.TurtleScreen._RUNNING = True
    screen = turtle.Screen()
    screen.clear()
    screen.colormode(255)
    screen.tracer(0)
    t = turtle.Turtle()
    t.speed(0)
    t.hideturtle()
    return screen, t
````

## Step 1: What is a function?

A **function** is a named block of code you can run over and over just by calling its name.

```python
def greet():
    print("Hello!")

greet()   # prints Hello!
greet()   # prints Hello! again
```

A **parameter** lets you pass information in:

```python
def greet(name):
    print("Hello,", name)

greet("Alice")   # Hello, Alice
greet("Bob")     # Hello, Bob
```

Our `bird` function will take a `radius` parameter so each bird can be a different size.

## Step 2: Drawing one arc with `turtle.circle()`

`t.circle(radius, extent)` draws an arc.
- `radius` controls how big the curve is
- `extent` is how many degrees of the circle to draw (90 = a quarter circle)
- `setheading(135)` points the turtle upper-left before each arc so the bird stays parallel to the horizon

Run this to see two identical arcs that form a bird shape:

````python
screen, t = fresh_turtle()

t.setheading(135)
t.circle(50, 90)
t.setheading(135)
t.circle(50, 90)

screen.update()
````

## Step 3: Drawing a few birds manually

Each bird is two identical arcs, both starting at heading 135. Let's draw three birds at different positions and sizes before we turn the pattern into a function:

````python
screen, t = fresh_turtle()

# bird 1
t.setheading(135)
t.circle(40, 90)
t.setheading(135)
t.circle(40, 90)

# bird 2 — moved over
t.penup()
t.goto(150, 50)
t.pendown()
t.setheading(135)
t.circle(25, 90)
t.setheading(135)
t.circle(25, 90)

# bird 3 — moved again
t.penup()
t.goto(-100, 80)
t.pendown()
t.setheading(135)
t.circle(60, 90)
t.setheading(135)
t.circle(60, 90)

screen.update()
````

## Step 4: Write the `bird` function

The repeated six lines above are begging to become a function. We pull them into `bird(radius)` so we can call it with any size:

````python
screen, t = fresh_turtle()

def bird(radius):
    t.setheading(135)
    t.circle(radius, 90)
    t.setheading(135)
    t.circle(radius, 90)

bird(40)

t.penup()
t.goto(150, 50)
t.pendown()
bird(25)

t.penup()
t.goto(-100, 80)
t.pendown()
bird(60)

screen.update()
````

## Step 5: Random numbers

To scatter birds across the screen we need random positions and random sizes. Python's `random` module gives us `randint(a, b)` which returns a random whole number between `a` and `b`.

Run this a few times and watch the numbers change:

````python
import random

print("random x:", random.randint(-300, 300))
print("random y:", random.randint(-200, 200))
print("random radius:", random.randint(20, 60))
````

## Step 6: Write the `birds` function

`birds(num_birds)` runs a loop `num_birds` times. Each iteration:
1. Moves the turtle to a random location
2. Calls `bird()` with a random radius

The turtle must be in `penup` before moving so it doesn't drag a line across the screen.

````python
import random

screen, t = fresh_turtle()

def bird(radius):
    t.setheading(135)
    t.circle(radius, 90)
    t.setheading(135)
    t.circle(radius, 90)

def birds(num_birds):
    for i in range(num_birds):
        t.penup()
        t.goto(random.randint(-300, 300), random.randint(-200, 200))
        t.pendown()
        bird(random.randint(20, 60))

birds(10)

screen.update()
````

## Step 8: Complete program

Here is the final clean version with both functions:

````python
import random

screen, t = fresh_turtle()

def bird(radius):
    t.setheading(135)
    t.circle(radius, 90)
    t.setheading(135)
    t.circle(radius, 90)

def birds(num_birds):
    for i in range(num_birds):
        t.penup()
        t.goto(random.randint(-300, 300), random.randint(-200, 200))
        t.pendown()
        bird(random.randint(20, 60))

birds(15)

screen.update()
````
