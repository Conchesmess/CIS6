# Lesson 3: For Loops

A **for loop** lets you run the same block of code a set number of times without copy-pasting it over and over.
By the end of this notebook you will use a loop to draw a solid filled square — one line at a time — without ever using the fill commands.

## Step 1: Your First For Loop

The basic pattern is:

```python
for i in range(n):
    # this runs n times
```

`range(5)` produces the numbers 0, 1, 2, 3, 4 — five numbers total.  
The variable `i` holds the current number each time through the loop.

Run the cell below and read the output.

````python
for i in range(5):
    print("Loop number:", i)
````

Notice that the loop starts at **0**, not 1. That is normal in Python.

You can also use `range(start, stop)` to control where the count begins:

````python
for i in range(1, 6):
    print("Loop number:", i)
````

## Step 2: Loops with Turtle — Drawing a Square

Remember from Lesson 1 how we drew a square by repeating the same two lines four times?

```python
t.forward(100)
t.right(90)
# ... repeated 4 times
```

A loop makes that much cleaner. Set up your turtle first:

````python
import turtle

screen = turtle.Screen()
screen.title("Loops")
screen.bgcolor("white")
screen.colormode(255)

t = turtle.Turtle()
t.speed(3)
````

````python
t.clear()
t.home()

for i in range(4):
    t.forward(100)
    t.right(90)
````

## Step 3: The speed() Command

`t.speed()` takes a number from **1** (slowest) to **10** (fastest).

| Value | Effect |
|---|---|
| 1 | Slowest — good for watching step by step |
| 3 | Slow |
| 6 | Normal |
| 10 | Fast |

Run the cells below and watch how the speed changes.

````python
t.clear()
t.home()
t.speed(1)          # slowest

for i in range(4):
    t.forward(100)
    t.right(90)
````

````python
t.clear()
t.home()
t.speed(10)         # fastest

for i in range(4):
    t.forward(100)
    t.right(90)
````

## Step 4: Drawing a Solid Square — One Line at a Time

Here is the challenge: fill a square with color **without** using `begin_fill()` or `end_fill()`.

The idea: draw 255 vertical lines, each one pixel apart, until you have a solid block of color.

```
| | | | | | |  <-- each | is one line, 255 pixels tall
```

Here is how each step of the loop works:
1. Draw a line straight up (255 pixels)
2. Lift the pen up
3. Move back down to the bottom
4. Shift one pixel to the right
5. Put the pen down — ready for the next line

Let's build it one piece at a time.

### Part A: Draw One Vertical Line

`t.setheading(90)` points the turtle straight up (90 degrees = north).

````python
t.clear()
t.penup()
t.goto(-127, -127)      # start at bottom-left of where the square will be
t.pendown()
t.setheading(90)        # face straight up

t.forward(255)          # draw one vertical line
````

### Part B: Move Over and Draw a Second Line

After drawing a line, we need to:
- Lift the pen so we don't draw while moving back down
- Use `t.goto()` to jump to the bottom of the next column (one pixel to the right)

````python
t.clear()
t.penup()
t.goto(-127, -127)
t.pendown()
t.setheading(90)

# Line 1
t.forward(255)

# Move to the bottom of the next column
t.penup()
t.goto(t.xcor() + 1, -127)   # t.xcor() is the turtle's current x position
t.pendown()

# Line 2
t.forward(255)
````

### Part C: Loop it 255 Times

Now wrap the pattern in a `for` loop that runs 255 times.  
Set `speed(10)` — you are about to ask the turtle to draw a lot of lines.

````python
t.clear()
t.speed(10)
t.penup()
t.goto(-127, -127)
t.pendown()
t.setheading(90)

for i in range(255):
    t.forward(255)                  # draw one vertical line
    t.penup()
    t.goto(t.xcor() + 1, -127)     # shift one pixel right, back to bottom
    t.pendown()
````

It works — but it is slow even at `speed(10)`. That is because `speed()` only controls how fast the turtle *animates*. The screen still has to redraw 255 times.

## Step 5: Making It Instantaneous with tracer()

The `screen` object has a command called `tracer()` that controls whether the screen updates after every move.

- `screen.tracer(0)` — turns off automatic screen updates entirely
- `screen.update()` — forces the screen to redraw once, right now

The turtle still does all the work — you just don't see it happening step by step. When you call `screen.update()` at the end, the finished drawing appears all at once.

````python
t.clear()
screen.tracer(0)            # turn off animation

t.penup()
t.goto(-127, -127)
t.pendown()
t.setheading(90)

for i in range(255):
    t.forward(255)
    t.penup()
    t.goto(t.xcor() + 1, -127)
    t.pendown()

screen.update()             # show the finished drawing all at once
````

---\n\n**Why draw 255 separate lines instead of just using `begin_fill()`?**\n\nBecause each line can be a *different color*. In Lesson 4 we will change the pen color on every iteration of this same loop — starting yellow and shifting to red one step at a time — to create a gradient that `begin_fill()` could never make.

The square appears instantly.

> **Important:** Once you use `tracer(0)`, you must always call `screen.update()` at the end of your drawing code or nothing will appear on screen.
